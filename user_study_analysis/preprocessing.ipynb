```python
import pandas as pd

# Read the previs_scores.csv file
previs_scores = pd.read_csv('previs_scores.csv')

# Display the first few rows of the dataframe
previs_scores.head()
```

```python
# compute the subscale averages
previs_scores['Understand'] = previs_scores[['Understand.1', 'Understand.2', 'Understand.3']].mean(axis=1).round(2)
previs_scores['Layout'] = previs_scores[['Layout.1', 'Layout.2', 'Layout.3']].mean(axis=1).round(2)
previs_scores['DataRead'] = previs_scores[['DataRead.1', 'DataRead.2', 'DataRead.3']].mean(axis=1).round(2)
previs_scores['DataFeat'] = previs_scores[['DataFeat.1', 'DataFeat.2']].mean(axis=1).round(2)

# Display the updated dataframe
previs_scores.head()

# drop the detailed columns
columns_to_drop = [
    'Understand.1', 'Understand.2', 'Understand.3',
    'Layout.1', 'Layout.2', 'Layout.3',
    'DataRead.1', 'DataRead.2', 'DataRead.3',
    'DataFeat.1', 'DataFeat.2'
]
previs_scores_averages = previs_scores.drop(columns=columns_to_drop)

# Display the dataframe with averages
previs_scores_averages.head()

# save the updated dataframe in CSV
previs_scores_averages.to_csv('previs_scores_averages.csv', index=False)
```

```python
# separate the data by visualization type
previs_scores_averages_matrix = previs_scores_averages[previs_scores_averages['Visualization'] == 'Matrix']
previs_scores_averages_similarity = previs_scores_averages[previs_scores_averages['Visualization'] == 'SimilarityView']

# compute the average of the subscales across all participants
previs_scores_averages_matrix_mean = previs_scores_averages_matrix[['Understand', 'Layout', 'DataRead', 'DataFeat']].mean().round(2)
previs_scores_averages_similarity_mean = previs_scores_averages_similarity[['Understand', 'Layout', 'DataRead', 'DataFeat']].mean().round(2)

# Display the mean scores for each visualization type
print("Matrix Visualization Mean Scores:")
print(previs_scores_averages_matrix_mean)
print("\nSimilarity View Visualization Mean Scores:")
print(previs_scores_averages_similarity_mean)

# save the mean scores in CSV
previs_scores_averages_matrix_mean.to_csv('previs_scores_averages_matrix_mean.csv', index=True)
previs_scores_averages_similarity_mean.to_csv('previs_scores_averages_similarity_mean.csv', index=True)
```
